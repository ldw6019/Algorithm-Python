## 1. 피보나치 수열
- $F(n) = F(n-1) + F(n-2)$, $F(0) = 0$, $F(1) = 1$
- Top-down: 재귀 + 메모이제이션
- Bottom-up: 반복문 + 테이블 채우기
- 시간 복잡도: O(n), 공간 복잡도: O(n) (메모이제이션) 또는 O(1) (반복문)

```python
 #top-down (하향식)

n = int(input())
fib = [None] * (n+1)

def top_fibo(n):
    if (fib[n] == None):
        if n < 2:
            fib[n] = n
        else:
            fib[n] = top_fibo(n-2) + top_fibo(n-1)
    return fib[n]
print(top_fibo(n))
```

```python
# bottom-up (상향식)

n = int(input())
def bottom_fibo(n):
    f = [None] * (n+1)
    f[0] = 0
    f[1] = 1
    for i in range(2, n+1):
        f[i] = f[i-2] + f[i-1]
    return f[n]
result = bottom_fibo(n)
print (result)
```

## 2. 이항계수 구하기

- 파스칼의 삼각형

  - C(n, k): n개의 원소에서 k개를 뽑는 경우의 수
  - C(n, 0) = C(n, n) = 1 **(n개에서 0개를 뽑는 경우의 수, n개에서 n개를 뽑는 경우의 수는 항상 1)**

- 순환 관계식
  - C(n, k) = C(n-1, k-1) + C(n-1, k) <br> **(n개에서 k개를 뽑는 경우의 수는 n-1개에서 k-1개를 뽑는 경우의 수 + n-1개에서 k개를 뽑는 경우의 수)**
- 시간 복잡도: $O(nk)$, 공간 복잡도: $O(nk)$

```python
# 이항계수 구하기(재귀, 많은 실행 발생함, 비효율)
n, k = map(int, input().split())
def triangle(n, k):
    if (k==0 or n==k):
        return 1
    return triangle(n-1, k-1) + triangle(n-1, k)

print(triangle(n, k))
```

```python
# 이항계수 구하기(동적 계획법, 메모이제이션)
n, k = map(int, input().split())
def table(n, k):
    dp = [[-1] * (k + 1) for _ in range(n + 1)]
    
    for i in range(n + 1):
        for j in range(min(i, k)+1):
            if j==0 or i==j:
                dp[i][j] = 1
            else:
                dp[i][j] = dp[i-1][j-1] + dp[i-1][j]
    return dp[n][k]
result = table(n, k)
print(result)
```

## **3. 0-1 Knapsack Problem**
- 배낭에 담을 수 있는 최대 가치 구하기
  - 가방 무게 W, 물건의 수 n
  - 각 물건의 무게 $w[i]$와 가치 $v[i]$
- $dp[i][j]$: i번째 물건까지 고려했을 때, j 무게
  - 점화식: $dp[i][j]$ = max(dp[i-1][j], dp[i-1][j-w[i]] + v[i]) if j >= w[i] else dp[i-1][j]
  - $A(n, W)$ = $dp[n][W]$ **(최대 가치)**

- **시간 복잡도: $O(nW)$**
### **중요도 높음+워낙 좃같으니 꼭 암기해서 시험장 가기**


```python
# W: 배낭의 최대 무게, wt: 물건의 무게 리스트, val: 물건의 가치 리스트, n: 물건의 수
def knapsack_dp(W, wt, val, n):
    A = [[0 for _ in range(W + 1)] for _ in range(n + 1)]

    for i in range(1, n+1):
        for w in range(1, W+1):
            # 현재 물건의 무게가 현재 배낭의 용량보다 크면, 이전 물건까지의 최대 가치를 그대로 가져옴
            if wt[i-1] > w:
                A[i][w] = A[i-1][w]

            # 그렇지 않으면, 현재 물건을 포함하는 경우와 포함하지 않는 경우 중에서 최대 가치를 선택
            else:
                valWith = val[i-1]+ A[i-1][w-wt[i-1]]
                valWithout = A[i-1][w]
                A[i][w] = max(valWith, valWithout)
    return A[n][W]
```

## 4. 최장 공통 부분 수열 (Longest Common Subsequence, LCS)
- 두 문자열 X와 Y의 최장 공통 부분 수열의 길이 구하기
  - $dp[i][j]$: X의 i번째 문자까지와 Y의 j번째 문자까지의 LCS 길이
  - 점화식:
    - if X[i] == Y[j]: $dp[i][j]$ = $dp[i-1][j-1] + 1$
    - else: $dp[i][j]$ = max($dp[i-1][j]$, $dp[i][j-1]$) 
  
<br>

- 최종 답: $dp[len(X)][len(Y)]$
- 시간 복잡도: O(mn), 공간 복잡도: O(mn) (m과 n은 각각 X와 Y의 길이)

```python
# table을 다 채운 후, 최장 공통 부분 수열을 구하는 백트래킹 함수
def backtracking(X, Y, table):
    lcs = ""
    i = len(X)
    j = len(Y)
    # table의 오른쪽 아래에서 시작하여, 최장 공통 부분 수열을 역추적
    while (i>0 and j>0):
        v = table[i][j]
        # LCS의 길이가 증가하는 경우, 즉 X[i-1]과 Y[j-1]이 같아서 LCS에 포함되는 경우
        if (v > table[i][j-1] and v > table[i-1][j] and v > table[i-1][j-1]):
            i -= 1
            j -= 1
            lcs = X[i] + lcs
            
        # LCS의 길이가 증가하지 않는 경우, 즉 X[i-1]과 Y[j-1]이 다르거나 LCS에 포함되지 않는 경우
        elif (v == table[i][j-1] and v > table[i-1][j]): j -= 1
        else: i -= 1
    return lcs

# 최장 공통 부분 수열의 길이를 구하는 동적 계획법 함수
def lcs_dp(X, Y):
    m = len(X)
    n = len(Y)
    table = [[None]*(n+1) for _ in  range(m+1)]
    
    for i in range(m+1):
        for j in range(n+1):
            # 초기화: 첫 번째 행과 첫 번째 열은 0으로 채움 (공통 부분 수열이 없으므로)
            if (i==0 or j==0): 
                table[i][j] = 0

            # X의 i-1번째 문자와 Y의 j-1번째 문자가 같으면, 이전까지의 LCS 길이에 1을 더함
            elif X[i-1] == Y[j-1]:
                table[i][j] = table[i-1][j-1] + 1

            # 그렇지 않으면, X의 i-1번째 문자까지의 LCS 길이와 Y의 j-1번째 문자까지의 LCS 길이 중에서 큰 값을 선택
            else:
                table[i][j] = max(table[i-1][j], table[i][j-1])
    
    # 테이블 출력 (디버깅용)
    for i in range(m+1):
        print(table[i])

    # 최장 공통 부분 수열을 구하는 백트래킹 함수 호출
    print("LCS = ", backtracking(X, Y, table))
    return table[m][n]
```

```python
X = "GAME OVER"

Y = "HELLO WORLD"

print("X =",X)
print("Y =", Y)
print("LCS(동적 계획)", lcs_dp(X , Y) )
```

## 5. 그래프 최단 경로 문제 (Floyd-Warshall Algorithm)
- 모든 쌍 최단 경로 알고리즘
  - $dp[k][i][j]$: k번째 정점까지 고려했을
  - 점화식: $dp[k][i][j]$ = min($dp[k-1][i][j]$, $dp[k-1][i][k]$ + $dp[k-1][k][j]$)
  - 최종 답: $dp[n][i][j]$ (n은 정점의 수)
- 시간 복잡도: $O(n^3)$, 공간 복잡도: $O(n^3)$
