# [Greedy Algorithm]
- 최적의 해를 구하기 위해 매 선택에서 가장 좋은 것을 선택하는 방법
- 대표적인 문제: 거스름돈 문제, 활동 선택 문제, 허프만 코딩 등
- 항상 최적의 해를 보장하지는 않음 (예: 0-1 Knapsack 문제)

## 1. 거스름돈 문제
- 동전의 종류와 거슬러 줄 금액이 주어졌을 때, 최소한의 동전 개수로 거슬러 주는 방법
- 동전의 종류가 1, 5, 6원일 때, 11원을 거슬러 주는 방법
- 그리디 알고리즘: 6원 1개 + 5원 1개 (총 2개)
- 하지만, 5원 2개 + 1원 1개 (총 3개)도 가능하므로, 그리디 알고리즘이 항상 최적의 해를 보장하지 않음

```python
def min_coins(coins, V):
    count = []
    for i in range(len(coins)):
        temp = V // coins[i]
        count.append(temp)
        V -= temp * coins[i]
    return count

coins = [500, 100, 50, 10, 5, 1]
V = 1280
print("동전의 종류:", coins)
print("거슬러 줄 금액:", V)
print("필요한 동전 개수:", min_coins(coins, V))
```

## 2. Fractional Knapsack Problem(분할 가능한 배낭 문제)
- 각 아이템의 무게와 가치가 주어졌을 때, 배낭의 최대 무게를 초과하지 않으면서 최대 가치를 얻는 방법
- 아이템을 분할할 수 있는 경우, 그리디 알고리즘이 최적의 해를 보장함
- 아이템을 가치/무게 비율로 정렬한 후, 가장 높은 비율부터 채워 나감(가성비)
- 0-1 Knapsack 문제에서는 아이템을 분할할 수 없으므로, 그리디 알고리즘이 최적의 해를 보장하지 않음
- 시간 복잡도: $O(n \log_2 n)$ (정렬)

```python
def knapsack_greedy(obj, W):
    # obj를 가치/무게 비율로 정렬 (가성비)
    obj.sort(key = lambda o: o[2]/o[1], reverse = True)

    totalValue = 0
    for o in obj:
        # 만약 배낭이 가득 찼다면 종료
        if W <= 0: break
        
        # 아이템을 통째로 넣을 수 있다면 넣기
        if W - o[1] >= 0:
            W -= o[1]
            totalValue += o[2]
            
        # 아이템을 분할해서 넣어야 한다면, 남은 배낭 무게에 맞게 분할해서 넣기
        else:
            fraction = W/o[1]
            totalValue += o[2] * fraction
            W = int(W- (o[1] * fraction))
    return totalValue
```

```python
# (물건,무게,가치)
obj = [ ("A", 10, 80), ("B", 12, 120), ("C", 8, 60)]
print("W = 18")
print("objects = ", obj)
print("부분적인배낭(18): ", knapsack_greedy(obj,18), end='\n\n')

obj= [("A", 10, 60), ("B", 40, 40), ("C", 20, 100), ("D", 30, 120) ]
print("W = 50")
print("objects = ", obj)
print("부분적인배낭(50): ", knapsack_greedy(obj, 50))
```

## **3. 최소 비용 신장 트리 (Minimum Spanning Tree, MST)**
- 그래프에서 모든 정점을 연결하는 트리 중에서 간선의 가중치 합이 최소인 트리를 찾는 문제
- 대표적인 알고리즘: **Prim, Kruskal\**
- Prim: 시작 정점에서 출발하여 가장 낮은 가중치의 간선을 선택하여 트리를 확장해 나감
- Kruskal: 모든 간선을 가중치 순으로 정렬한 후, 사이클이 생기지 않도록 간선을 선택하여 트리를 구성함

<br>

- 시간 복잡도: 
  - Prim - $O(E \log_2 V)$
  - Kruskal - $O(E \log_2 E)$ 
- 최적의 해를 보장함


---
### 3-1. Prim 알고리즘
- 시작 정점에서 출발하여 가장 낮은 가중치의 간선을 선택하여 트리를 확장해 나감
<br>

- 진행 흐름:
  - 시작 정점 선택
  - 현재 트리에 연결된 간선 중에서 가장 낮은 가중치의 간선을 선택하여 트리에 추가
  - 새로운 정점이 트리에 추가되면, 그 정점과 연결된 간선들을 고려하여 다음 간선을 선택
  - 모든 정점이 트리에 포함될 때까지 반복

```python
INF = int(1000)

# 최소 비용 정점의 번호를 반환하는 함수
def getMinVertex(dist, selected):
    # minv: 최소 비용 정점의 번호, mindist: 최소 비용
    minv = -1
    mindist = INF

    
    for v in range(len(dist)) :
        # 선택되지 않은 정점 중에서 최소 비용인 정점 v를 찾음
        if not selected[v] and dist[v]<mindist :
            mindist = dist[v]
            minv = v
    return minv


def MSTPrim(vertex, adj):
    # vertex: 정점 리스트, adj: 인접 행렬, dist: 최소 비용 리스트, selected: 선택된 정점 리스트
    vsize = len(vertex)
    dist = [INF] * vsize
    dist[0] = 0
    selected = [False] * vsize

    # MST를 구성하는 vsize-1개의 간선을 선택
    for i in range(vsize):
        # 선택되지 않은 정점 중에서 최소 비용인 정점 u를 선택
        u = getMinVertex(dist, selected)
        selected[u] = True
        # 출력: 선택된 정점과 그 비용
        print(vertex[u], end=':')
        print(dist)

        for v in range(vsize) :
            if (adj[u][v] != None):
                if selected[v] == False and adj[u][v]< dist[v]:
                    # u를 거쳐서 v로 가는 비용이 현재 dist[v]보다 작은 경우, dist[v]를 adj[u][v]로 갱신
                    dist[v] = adj[u][v] 
```

```python
vertex = [ 'A', 'B', 'C', 'D', 'E', 'F', 'G' ]
weight = [ [0,   7,   INF, INF, 3,   10,  INF],
          [ 7,   0,   4,   10,  2,   6,   INF],
          [ INF, 4,   0,   2,   INF, INF, INF],
          [ INF, 10,  2,   0,   11,  INF, 9],
          [ 3,   2,   INF, 11,  0,   13,  5],
          [ 10,  6,   INF, 9,   13,  0,   INF],
          [ INF, INF, INF, 4,   5,  INF,  0]]

print("MST By Prim's Algorithm")
MSTPrim(vertex, weight)

```

### 3-2. Kruskal 알고리즘
- 모든 간선을 가중치 순으로 정렬한 후, `사이클이 생기지 않도록` 간선을 선택하여 트리를 구성함
<br>
- 진행 흐름:
  - 모든 간선을 가중치 순으로 정렬
  - `가장 낮은 가중치`의 간선부터 선택하여 트리에 추가
  - 간선을 추가할 때, 사이클이 생기지 않도록 **`유니온-파인드(Union-Find)`** 자료구조를 사용하여 연결 여부를 확인
  - 모든 정점이 트리에 포함될 때까지 반복
  - 최적의 해를 보장함

<br>

- 시간 복잡도:
  - 그래프가 인접 행렬: $O(n^2 + e\log_2 e)$
  - 그래프가 인접 리스트: $O(e \log_2 e)$ 
  

```python
# 수업자료 기반 union-find 자료구조 구현
class disjointSets:
    def __init__(self, n):
        self.parent = [-1]*n                # 각 노드는 모두 루트 노드
        self.set_size = n                   # 전체 집합의 개수

    def find(self, id):                     # 정점 id가 속한 집합의 대표번호 탐색
        while (self.parent[id] >= 0):       # 부모가 있으면(-1이 아닌 동안)
            id = self.parent[id]            # id를 부모 id로 갱신
        return id                           # 최종 id 반환. 루트 노드의 id임

    def union(self, s1, s2):                # 두 집합을 병합(s1을 s2에 병합시킴)
        self.parent[s1] = s2                # s1을 s2에 병합시킴
        self.set_size = self.set_size-1     # 집합의 개수가 줄어 듦
```

```python
def MSTKruskal(V, adj):
    n = len(V)
    ds = disjointSets(n)
    E = []

    for i in range(n-1):
        for j in range(i+1, n):
            if adj[i][j] != None:
                E.append((i, j, adj[i][j]))

    E.sort(key=lambda e:e[2])

    ecount = 0
    for i in range(len(E)):
        e = E[i]
        uset = ds.find(e[0])
        vset = ds.find(e[1])

        if uset != vset:
            ds.union(uset, vset)
            print("간선 추가: (%s, %s, %d)" %(V[e[0]], V[e[1]], e[2]))
            ecount += 1
            if ecount == (n-1):
                break
```

```python
vertex = [ 'A', 'B', 'C', 'D', 'E', 'F', 'G' ]
weight = [ [0,   7,   INF, INF, 3,   10,  INF],
          [ 7,   0,   4,   10,  2,   6,   INF],
          [ INF, 4,   0,   2,   INF, INF, INF],
          [ INF, 10,  2,   0,   11,  INF, 9],
          [ 3,   2,   INF, 11,  0,   13,  5],
          [ 10,  6,   INF, 9,   13,  0,   INF],
          [ INF, INF, INF, 4,   5,  INF,  0]]

print("MST By Kruskal's Algorithm")
MSTKruskal(vertex, weight)
```

### **4. Dijkstra 알고리즘**
- `단일 출발점`에서 모든 다른 정점까지의 최단 경로를 찾는 알고리즘
- 시간 복잡도:
  - $O(V^2)$ (인접 행렬)
  - $O((V + E) \log_2 V)$ (인접 리스트 + 우선순위 큐)
- 최적의 해를 보장함
- 음수 간선이 있는 경우에는 사용할 수 없음

```python
# Prim에서 사용했던 최소 비용 정점의 번호를 반환하는 함수
def getMinVertex(dist, selected):
    # minv: 최소 비용 정점의 번호, mindist: 최소 비용
    minv = -1
    mindist = INF
    for v in range(len(dist)) :
        # 선택되지 않은 정점 중에서 최소 비용인 정점 v를 찾음
        if not selected[v] and dist[v]<mindist :
            mindist = dist[v]
            minv = v
    return minv

# vtx: 정점 리스트, adj: 인접 행렬, start: 시작 정점의 번호
def dijkstra(vtx, adj, start):
    vsize = len(vtx)
    dist = list(adj[start])
    dist[start] = 0
    
    # path: 최단 경로 트리를 구성하는 부모 정점 리스트, found: 최단 경로 트리에 포함된 정점 리스트
    path = [start] * vsize
    found = [False] * vsize
    found[start] = True


    for i in range(vsize):
        # 선택되지 않은 정점 중에서 최소 비용인 정점 u를 선택
        print("Step%2d: " %(i+1), dist)
        u = getMinVertex(dist, found)
        found[u] = True

        for w in range(vsize):
            # 선택되지 않은 정점 w에 대해서, u를 거쳐서 w로 가는 비용이 현재 dist[w]보다 작은 경우, dist[w]를 갱신
            if not found[w]:
                if dist[u] + adj[u][w] < dist[w]:
                    dist[w] = dist[u] + adj[u][w]
                    path[w] = u
    return path        
```

```python
INF = 9999
vertex = [ 'A', 'B', 'C', 'D', 'E', 'F', 'G' ]
weight = [ [   0,   7, INF, INF,   3,  10, INF ],
           [   7,   0,   4,  10,   2,   6, INF ],
           [ INF,   4,   0,   2, INF, INF, INF ],
           [ INF,  10,   2,   0,  11,   9,   4 ],
           [   3,   2, INF,  11,   0,  13,   5 ],
           [  10,   6, INF,   9,  13,   0, INF ],
           [ INF, INF, INF,   4,   5, INF,   0 ]]

print("Shortest Path By Dijkstra Algorithm")
start = 0   # 시작 정점은 0번(A)
path = dijkstra(vertex, weight, start)
```

```python
# 최종 경로를 출력하기 위한 코드
for end in range(len(vertex)) :

    # start에서 end로 가는 최단 경로를 출력
    if (end != start) :
        print("[최단경로: %s -> %s] %s" %
        (vertex[start], vertex[end], vertex[end]), end='')

        # path[end]는 end 정점의 부모 정점 번호를 나타냄. path[end]가 start가 될 때까지 반복하면서 경로를 역추적하여 출력
        while (path[end] != start) :
            print(" <- %s" % vertex[path[end]], end='')
            end = path[end]

        print(" <- %s" % vertex[path[end]])
```

## 5. Huffman Coding
- 가변 길이의 이진 코드를 사용하여 데이터를 압축하는 알고리즘
- 각 문자의 빈도에 따라 가중치를 부여하여 트리를 구성
- 빈도가 높은 문자는 짧은 코드, 빈도가 낮은 문자는 긴 코드를 할당하여 전체 데이터의 크기를 줄임
- 시간 복잡도: $O(n \log_2 n)$ (우선순위 큐 사용)
- 최적의 해를 보장함

```python
import heapq

def make_heap_tree(freq, label):
    n = len(freq)
    h = []
    for i in range(n):
        heapq.heappush(h, (freq[i], label[i]))

    for i in range(1, n):
        e1 = heapq.heappop(h)
        e2 = heapq.heappop(h)
        heapq.heappush(h, (e1[0]+e2[0], e1[1]+e2[1]))
        print(e1, "+", e2)

    print(heapq.heappop(h))
```

```python
label = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']
freq  = [24,    3,   8,  10,  33,   6,   4,  12]
make_heap_tree(freq, label)
```
