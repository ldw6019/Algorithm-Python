
# [Backtracking & Branch and Bound]
- **분기한정은 범위에서 제외됨**
  
  <br>
- 백트래킹은 해가 될 수 없는 경우를 미리 판단하여 탐색을 중단하는 방법
- 분기한정은 해가 될 수 없는 경우를 미리 판단하여 탐색을 중단하는 방법과 함께, 해가 될 수 있는 경우에도 탐색을 중단하여 최적해를 찾는 방법
- 백트래킹은 해가 될 수 없는 경우를 미리 판단하여 탐색을 중단하는 방법이지만, 분기한정은 해가 될 수 없는 경우를 미리 판단하여 탐색을 중단하는 방법과 함께, 해가 될 수 있는 경우에도 탐색을 중단하여 최적해를 찾는 방법입니다.


---
## 1. 간단한 문제 해결 (순열 생성)
- 백트래킹을 사용하여 주어진 리스트의 모든 순열을 생성하는 예제
- 시간 복잡도: $O(n*n!)$ (모든 순열을 생성하기 때문)
- 최적의 해를 보장함 (모든 가능한 순열을 탐색하기 때문)
